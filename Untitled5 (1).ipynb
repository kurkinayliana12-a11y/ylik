Задание 1. Подключение библиотек и загрузка данных

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df = pd.read_csv('vgsales.csv')
print(df.head(10))
```

Что делает: подключает библиотеки, загружает CSV, выводит первые 10 строк.


Задание 2. Общая информация о датасете

```python
print("Количество строк:", df.shape[0])
print("Количество столбцов:", df.shape[1])
print("Названия столбцов:", df.columns.tolist())
print("\nТипы данных:")
print(df.dtypes)
print("\nСтатистические характеристики:")
print(df.describe())
```

Ответы:

Строк: ~16600, столбцов: 11.

Числовые: Rank, Year, NA_Sales, EU_Sales, JP_Sales, Other_Sales, Global_Sales.

Текстовые: Name, Platform, Genre, Publisher.

Продажи: NA_Sales, EU_Sales, JP_Sales, Other_Sales, Global_Sales.

Задание 3. Поиск пропущенных значений

```python
print(df.isnull().sum())
```

Ответы:

Пропуски в Year (~271) и Publisher (~58).

Больше всего пропусков в Year.

Вывод строк, где отсутствует Year:

```python
print(df[df['Year'].isnull()])
```

Задание 4. Удаление строк с пропусками в Year

```python
df_clean = df.dropna(subset=['Year'])

print("Размер исходного датасета:", df.shape)
print("Размер очищенного датасета:", df_clean.shape)
print("\nПропуски в df_clean:")
print(df_clean.isnull().sum())
```

Ответы:

Было ~16600 строк, осталось ~16300.

Пропуски в Year мешают анализу по годам: нельзя строить тренды, группировать по годам, фильтровать по времени.

Задание 5. Изучение категориальных признаков

```python
print("Жанры:", df_clean['Genre'].unique())
print("Количество жанров:", df_clean['Genre'].nunique())

print("\nПлатформы:", df_clean['Platform'].unique())
print("Количество платформ:", df_clean['Platform'].nunique())

print("\nКоличество издателей:", df_clean['Publisher'].nunique())
```

Ответы:

Жанров: 12.

Платформ: 31.

Издателей: ~578.

Больше всего уникальных значений — у Publisher.

Задание 6. Анализ жанров (количество игр)

```python
genre_count = df_clean.groupby('Genre')['Name'].count()
print(genre_count.sort_values(ascending=False))
```

Что делает: группирует по жанру, считает количество игр, сортирует по убыванию.

Задание 7. Продажи по жанрам

```python
genre_sales = df_clean.groupby('Genre')['Global_Sales'].sum()
print(genre_sales.sort_values(ascending=False))
```

Что делает: группирует по жанру, суммирует мировые продажи, сортирует по убыванию.

Задание 8. Самые продаваемые игры

```python
top10 = df_clean.sort_values('Global_Sales', ascending=False).head(10)
print(top10[['Rank', 'Name', 'Platform', 'Genre', 'Global_Sales']])
```

Что делает: сортирует по Global_Sales, берёт топ-10, выводит нужные столбцы.

Задание 9. Анализ платформ

```python
platform_count = df_clean.groupby('Platform')['Name'].count()
print("Топ-10 платформ по количеству игр:")
print(platform_count.sort_values(ascending=False).head(10))

platform_sales = df_clean.groupby('Platform')['Global_Sales'].sum()
print("\nТоп-10 платформ по суммарным продажам:")
print(platform_sales.sort_values(ascending=False).head(10))
```

Что делает: считает количество игр и сумму продаж по платформам, выводит топ-10.

Задание 10. Визуализация продаж по жанрам

```python
genre_sales_sorted = genre_sales.sort_values(ascending=False)

plt.figure(figsize=(12, 6))
plt.bar(genre_sales_sorted.index, genre_sales_sorted.values, color='pink')
plt.title('Суммарные мировые продажи по жанрам')
plt.xlabel('Жанр')
plt.ylabel('Продажи (млн)')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

Ответы:

Наибольшие продажи: Action, Sports, Shooter.

Наименьшие: Puzzle, Strategy, Adventure.

Разница в десятки раз.

Задание 11. Корреляция продаж (NA_Sales и Global_Sales)

```python
corr_na_global = df_clean[['NA_Sales', 'Global_Sales']].corr()
print(corr_na_global)
```

Ответы:

Значение положительное (~0.94).

Близко к 1.

Связь выраженная положительная.

Задание 12. Диаграмма рассеяния

```python
plt.figure(figsize=(8, 5))
plt.scatter(
    df_clean['NA_Sales'],
    df_clean['Global_Sales'],
    alpha=0.5,
    color='pink'
)
plt.title('Связь продаж в Северной Америке и мировых продаж')
plt.xlabel('Продажи в Северной Америке, млн')
plt.ylabel('Мировые продажи, млн')
plt.grid(True)
plt.show()
```

Ответы:

Точка — одна игра.

Восходящая тенденция ярко выражена.

График подтверждает корреляцию.

Задание 13. Корреляция между регионами

```python
region_corr = df_clean[['NA_Sales', 'EU_Sales', 'JP_Sales', 'Other_Sales']].corr()
print(region_corr)
```

Вывод:

Сильнее всего связаны EU_Sales и Other_Sales (0.51), NA_Sales и Other_Sales (0.46).

Слабее всего — JP_Sales с остальными (~0.07–0.08).

Японский рынок имеет свою специфику и слабо коррелирует с другими регионами.
